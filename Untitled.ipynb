# Modeling 


### Loading Data 

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("data")

team_seasons = pd.read_csv(DATA_DIR / "team_seasons.csv")
analysis = pd.read_csv(DATA_DIR / "analysis.csv")

GROUPS = ["sustained_losing", "comparison"]

LABELS = {
    "sustained_losing": "Sustained losing",
    "comparison": "Comparison",
}

COLORS = {
    "sustained_losing": "#D55E00",
    "comparison": "#0072B2",
}

plt.rcParams.update({
    "figure.figsize": (9, 5),
    "font.size": 11,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
```

### Missing Values

```python
display(
    team_seasons.isna().sum().to_frame("Missing values")
)

display(
    analysis.isna().sum().to_frame("Missing values")
)

print(
    "Duplicate team-season rows:",
    team_seasons.duplicated(["team_id", "season_start"]).sum()
)

print(
    "Rows with inconsistent game totals:",
    (
        team_seasons["wins"] + team_seasons["losses"]
        != team_seasons["games"]
    ).sum()
)

coverage = team_seasons.groupby("season").agg(
    teams=("team_id", "nunique"),
    min_games=("games", "min"),
    max_games=("games", "max"),
)

display(coverage)
```

### Summarizing Groups and Counting Groups

```python
summary = (
    analysis.groupby("group")
    .agg(
        examples=("team_id", "size"),
        unique_teams=("team_id", "nunique"),
        baseline_win_pct=("qualifying_avg_win_pct", "mean"),
        future_mean_win_pct=("future_avg_win_pct", "mean"),
        future_median_win_pct=("future_avg_win_pct", "median"),
        future_win_pct_std=("future_avg_win_pct", "std"),
        avg_playoff_appearances=("future_playoff_appearances", "mean"),
        share_reaching_playoffs=("made_playoffs_within_5", "mean"),
    )
    .reindex(GROUPS)
    .rename(index=LABELS)
)

display(summary.round(3))

counts = analysis["group"].value_counts().reindex(GROUPS, fill_value=0)

fig, ax = plt.subplots()

bars = ax.bar(
    [LABELS[g] for g in GROUPS],
    counts.values,
    color=[COLORS[g] for g in GROUPS],
)

ax.bar_label(bars, padding=3)
ax.set(
    title="Number of qualifying examples by group",
    ylabel="Qualifying examples",
)
ax.set_ylim(0, max(1, counts.max()) * 1.2)

plt.tight_layout()
plt.show()
```

### Distribution of historical winning percentages

```python
fig, ax = plt.subplots()

ax.hist(
    team_seasons["win_pct"].dropna(),
    bins=20,
    color="#666666",
    edgecolor="white",
)

ax.axvline(
    0.350,
    color=COLORS["sustained_losing"],
    linestyle="--",
    label="Losing cutoff: .350",
)

ax.axvspan(
    0.400,
    0.550,
    color=COLORS["comparison"],
    alpha=0.15,
    label="Comparison range: .400–.550",
)

ax.set(
    title="Distribution of NBA team-season winning percentages",
    xlabel="Regular-season winning percentage",
    ylabel="Team-seasons",
    xlim=(0, 1),
)

ax.legend()
plt.tight_layout()
plt.show()
```

### Compare future winning percentages

```python
values = [
    analysis.loc[
        analysis["group"].eq(g), "future_avg_win_pct"
    ].dropna().to_numpy()
    for g in GROUPS
]

if any(len(v) == 0 for v in values):
    raise ValueError("Both groups need observations for this comparison.")

fig, ax = plt.subplots()

boxes = ax.boxplot(
    values,
    patch_artist=True,
    medianprops={"color": "black"},
)

for box, group in zip(boxes["boxes"], GROUPS):
    box.set_facecolor(COLORS[group])
    box.set_alpha(0.65)

ax.set_xticks([1, 2])
ax.set_xticklabels([LABELS[g] for g in GROUPS])

ax.axhline(0.500, color="gray", linestyle="--", label=".500 record")

ax.set(
    title="Future five-season winning percentage by group",
    ylabel="Average winning percentage over the next five seasons",
    ylim=(0, 1),
)

ax.legend()
plt.tight_layout()
plt.show()
```

### Track performance before and after the qualifying pair

```python
trajectory_columns = [
    "previous_win_pct",
    "anchor_win_pct",
] + [
    f"year_{year}_win_pct" for year in range(1, 6)
]

relative_years = [-1, 0, 1, 2, 3, 4, 5]

fig, ax = plt.subplots()

for group in GROUPS:
    group_data = analysis.loc[analysis["group"].eq(group)]
    means = group_data[trajectory_columns].mean()

    ax.plot(
        relative_years,
        means.to_numpy(),
        marker="o",
        linewidth=2,
        color=COLORS[group],
        label=f"{LABELS[group]} (n={len(group_data)})",
    )

ax.axvline(0.5, color="gray", linestyle="--")
ax.axhline(0.500, color="gray", linestyle=":", alpha=0.7)

ax.set_xticks(relative_years)
ax.set(
    title="Average team performance across the study window",
    xlabel="Season relative to anchor: −1 and 0 are qualifying seasons",
    ylabel="Mean regular-season winning percentage",
    ylim=(0, 1),
)

ax.legend()
plt.tight_layout()
plt.show()
```

### Compare playoff outcomes

```python
playoff_summary = (
    analysis.groupby("group")
    .agg(
        any_playoffs=("made_playoffs_within_5", "mean"),
        appearances=("future_playoff_appearances", "mean"),
    )
    .reindex(GROUPS)
)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

labels = [LABELS[g] for g in GROUPS]
colors = [COLORS[g] for g in GROUPS]

bars = axes[0].bar(
    labels,
    playoff_summary["any_playoffs"] * 100,
    color=colors,
)

axes[0].bar_label(bars, fmt="%.1f%%", padding=3)
axes[0].set(
    title="Reached the playoffs at least once",
    ylabel="Percentage of examples",
    ylim=(0, 110),
)

bars = axes[1].bar(
    labels,
    playoff_summary["appearances"],
    color=colors,
)

axes[1].bar_label(bars, fmt="%.2f", padding=3)
axes[1].set(
    title="Average playoff appearances",
    ylabel="Appearances over five seasons",
    ylim=(0, 5.5),
)

plt.tight_layout()
plt.show()
```

### Check whether a few teams dominate the sample

```python
team_counts = (
    analysis.groupby(["team_id", "group"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=GROUPS, fill_value=0)
)

team_names = (
    team_seasons.sort_values("season_start")
    .drop_duplicates("team_id", keep="last")
    .set_index("team_id")["team_name"]
)

team_counts.index = [
    f"{team_names.get(team_id, 'Unknown')} ({team_id})"
    for team_id in team_counts.index
]

team_counts = team_counts.loc[
    team_counts.sum(axis=1).sort_values().index
]

ax = team_counts.rename(columns=LABELS).plot.barh(
    stacked=True,
    figsize=(10, max(5, len(team_counts) * 0.3)),
    color=[COLORS[g] for g in GROUPS],
)

ax.set(
    title="Qualifying examples contributed by each team ID",
    xlabel="Number of examples",
    ylabel="Latest available team name and ID",
)

plt.tight_layout()
plt.show()
```
